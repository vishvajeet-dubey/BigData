#### 3. Difference between append(), extend() and insert()
- append() adds one object.
- extend() adds elements from an iterable.
- insert(index, value) inserts at a specific position

```python
# append()
data = [1,2,3,4]
data.append([5,6])
data

# append() adds one object.
```

```python
# extend()
data = [1,2,3,4]
data.extend([5,6])
data

# extend() adds elements from an iterable.
```

```python
# insert
data = [1,2,3,4]
data.insert(0, 5)
data

# insert(index, value) inserts at a specific position.
```

#### 4. Difference between remove(), pop() and del
- remove() -> by value
- pop() -> remove by index and return the value
- del -> deletes an element, slice, or even the entire variable:

```python
# remove()
data = [1,2,3,4]
data.remove(1)
data

# remove() -> by value
```

```python
# pop()
data = [1,2,3,4]
data.pop(3)
data

# pop() -> remove by index and return the value
```

```python
# delete()
data = [1,2,3,4]
del data[1]
data

# del -> deletes an element, slice, or even the entire variable:
```

```python
# delete()
data = [1,2,3,4,5,6,7,8,9]
del data[::2]
data

# del -> deletes an element, slice, or even the entire variable:
```

#### 6. find the duplicate from the list

```python
data = [1,2,3,4,4,5,5,6,7,7]

def dupsFinder(data: list) -> set:
	seen = set()
	duplicate = set()

	for x in data:
		if x in seen:
			duplicate.add(x)
		else:
			seen.add(x)
	return duplicate

dupsFinder(data)
```

#### 7. find the duplicate without using set

```python
data = [1,2,3,4,4,5,5,6,7,7]

def dupsFinder(data:list) -> list:
    seen = list()
    duplicate = list()
    for x in data:
        if x in seen:
            if x in duplicate:
                continue
            else:
                duplicate.append(x)
        else:
            seen.append(x)
    return duplicate

dupsFinder(data)
```

#### 8. find the second highest number from the list

```python
data = [1,2,3,4,4,5,5,6,7,7]

def secondHigh(data: list)->int:
    # data = list(set(data))
    data.sort(reverse=True)
    return data[1]

secondHigh(data)
    
```

#### 9. Remove duplicates while preserving order

```python
data = [1,2,3,4,4,5,5,6,7,7]

def removeDups(data:list) -> list:
    uniqueList = []
    for i in data:
        if i in uniqueList:
            continue
        else:
            uniqueList.append(i)
    return uniqueList

removeDups(data)
```

#### 10. Find frequency of each element

```python
fruits = ['apple', 'banana', 'apple', 1,1,2,3,3]

def frequency(data:list)-> dict:
    freq = {}
    for x in data:
        freq[x]=data.count(x)
    return freq

frequency(fruits)
```

#### 11. Find common elements between two lists

```python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

common = []
for x in a:
    if x in b:
        common.append(x)

common
```

#### 12. Find elements present in one list but not another

```python
a = [1, 2, 3, 4]
b = [3, 4, 5]

result = []
for x in a:
    if x in b:
        continue
    else:
        result.append(x)

result
```

#### 13. Flatten a nested list

```python
data = [[1, 2, 3], [3, 4], [5, 6]]

result = []

for sublist in data:
    for x in sublist:
        result.append(x)

result
```

#### 14. Split a list into chunks of size n

```python
data = [1,2,3,4,5,6,7,8,9]
n = 4

result = []

for i in range(0, len(data), n):
    accumlator = data[i:i+n]
    result.append(accumlator)

result
```

```python
# using list comprehension
[data[i: i+n] for i in range(0, len(data), n)]
```

#### 15. Find maximum and minimum without max() and min()

```python
data = [1,2,3,4,5,0,7,8]

min = data[0]
max = data[0]

for i in data:
    if i < min:
        min=i
    elif i > max:
        max = i
    else:
        continue

print(f'min = {min} and max = {max}')
```
